```python
import numpy as np
import pandas as pd
```

```python
corpus = "bgsner"
bgsner_metrics_albert = pd.read_csv("bgsner/albert-base-v2_overall_metrics.csv")
name = "albert"
bgsner_metrics_albert_selected = bgsner_metrics_albert[
    ["split", "overall_accuracy", "f1_macro"]
].rename(
    columns={
        "overall_accuracy": f"overall_accuracy_{name}_{corpus}",
        "f1_macro": f"f1_macro_{name}_{corpus}",
    }
)

bgsner_metrics_bert = pd.read_csv(
    "bgsner/bert-base-cased_overall_metrics.csv"
)
name = "bert"
bgsner_metrics_bert_selected = bgsner_metrics_bert[
    ["split", "overall_accuracy", "f1_macro"]
].rename(
    columns={
        "overall_accuracy": f"overall_accuracy_{name}_{corpus}",
        "f1_macro": f"f1_macro_{name}_{corpus}",
    }
)

bgsner_metrics_roberta = pd.read_csv(
    "bgsner/xlm-roberta-base_overall_metrics.csv"
)
name = "roberta"
bgsner_metrics_roberta_selected = bgsner_metrics_roberta[
    ["split", "overall_accuracy", "f1_macro"]
].rename(
    columns={
        "overall_accuracy": f"overall_accuracy_{name}_{corpus}",
        "f1_macro": f"f1_macro_{name}_{corpus}",
    }
)

merged_bgsner = bgsner_metrics_bert_selected.merge(
    bgsner_metrics_albert_selected, on="split", how="outer"
).merge(bgsner_metrics_roberta_selected, on="split", how="outer")
```

```python
corpus = 'geocorpus'
geocorpus_metrics_albert = pd.read_csv("geocorpus/albertina-base_overall_metrics.csv")
name = "albert"
geocorpus_metrics_albert_selected = geocorpus_metrics_albert[
    ["split", "overall_accuracy", "f1_macro"]
].rename(
    columns={
        "overall_accuracy": f"overall_accuracy_{name}_{corpus}",
        "f1_macro": f"f1_macro_{name}_{corpus}",
    }
)

geocorpus_metrics_bert = pd.read_csv(
    "geocorpus/bert-base-portuguese-cased_overall_metrics.csv"
)
name = "bert"
geocorpus_metrics_bert_selected = geocorpus_metrics_bert[
    ["split", "overall_accuracy", "f1_macro"]
].rename(
    columns={
        "overall_accuracy": f"overall_accuracy_{name}_{corpus}",
        "f1_macro": f"f1_macro_{name}_{corpus}",
    }
)

geocorpus_metrics_roberta = pd.read_csv(
    "geocorpus/xlm-roberta-base_overall_metrics.csv"
)
name = "roberta"
geocorpus_metrics_roberta_selected = geocorpus_metrics_roberta[
    ["split", "overall_accuracy", "f1_macro"]
].rename(
    columns={
        "overall_accuracy": f"overall_accuracy_{name}_{corpus}",
        "f1_macro": f"f1_macro_{name}_{corpus}",
    }
)
```

```python
merged_geocorpus = geocorpus_metrics_bert_selected.merge(
    geocorpus_metrics_albert_selected, on="split", how="outer"
).merge(geocorpus_metrics_roberta_selected, on="split", how="outer")
```

```python
merged = merged_bgsner.merge(
    merged_geocorpus, on="split", how="outer")
```

```python
merged
```

```python
merged.columns
```

```python
split_order = ["std", "heur_len", "heur_rare", "advs", "loc", "semantic", "reverse"]

merged["split"] = merged["split"].astype(str).str.strip()

merged["split"] = pd.Categorical(merged["split"], categories=split_order, ordered=True)
merged = merged.sort_values("split", kind="stable").reset_index(
    drop=True
)
```

```python
accuracy_df = merged[["split"] + [c for c in merged.columns if "accuracy" in c]]
```
