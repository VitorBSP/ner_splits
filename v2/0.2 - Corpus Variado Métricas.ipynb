```python
import numpy as np
import pandas as pd
```

```python
corpus = "cleanconll"
cleanconll_metrics_albert = pd.read_csv("cleanconll/albert-base-v2_overall_metrics.csv")
name = "albert"
cleanconll_metrics_albert_selected = cleanconll_metrics_albert[
    ["split", "overall_accuracy", "f1_macro"]
].rename(
    columns={
        "overall_accuracy": f"overall_accuracy_{name}_{corpus}",
        "f1_macro": f"f1_macro_{name}_{corpus}",
    }
)

cleanconll_metrics_bert = pd.read_csv(
    "cleanconll/bert-base-cased_overall_metrics.csv"
)
name = "bert"
cleanconll_metrics_bert_selected = cleanconll_metrics_bert[
    ["split", "overall_accuracy", "f1_macro"]
].rename(
    columns={
        "overall_accuracy": f"overall_accuracy_{name}_{corpus}",
        "f1_macro": f"f1_macro_{name}_{corpus}",
    }
)

cleanconll_metrics_roberta = pd.read_csv(
    "cleanconll/xlm-roberta-base_overall_metrics.csv"
)
name = "roberta"
cleanconll_metrics_roberta_selected = cleanconll_metrics_roberta[
    ["split", "overall_accuracy", "f1_macro"]
].rename(
    columns={
        "overall_accuracy": f"overall_accuracy_{name}_{corpus}",
        "f1_macro": f"f1_macro_{name}_{corpus}",
    }
)

merged_cleanconll = cleanconll_metrics_bert_selected.merge(
    cleanconll_metrics_albert_selected, on="split", how="outer"
).merge(cleanconll_metrics_roberta_selected, on="split", how="outer")
```

```python
corpus = 'cachaca'
cachaca_metrics_albert = pd.read_csv("cachaca/albertina-base_overall_metrics.csv")
name = "albert"
cachaca_metrics_albert_selected = cachaca_metrics_albert[
    ["split", "overall_accuracy", "f1_macro"]
].rename(
    columns={
        "overall_accuracy": f"overall_accuracy_{name}_{corpus}",
        "f1_macro": f"f1_macro_{name}_{corpus}",
    }
)

cachaca_metrics_bert = pd.read_csv(
    "cachaca/bert-base-portuguese-cased_overall_metrics.csv"
)
name = "bert"
cachaca_metrics_bert_selected = cachaca_metrics_bert[
    ["split", "overall_accuracy", "f1_macro"]
].rename(
    columns={
        "overall_accuracy": f"overall_accuracy_{name}_{corpus}",
        "f1_macro": f"f1_macro_{name}_{corpus}",
    }
)

cachaca_metrics_roberta = pd.read_csv(
    "cachaca/xlm-roberta-base_overall_metrics.csv"
)
name = "roberta"
cachaca_metrics_roberta_selected = cachaca_metrics_roberta[
    ["split", "overall_accuracy", "f1_macro"]
].rename(
    columns={
        "overall_accuracy": f"overall_accuracy_{name}_{corpus}",
        "f1_macro": f"f1_macro_{name}_{corpus}",
    }
)

merged_cachaca = cachaca_metrics_bert_selected.merge(
    cachaca_metrics_albert_selected, on="split", how="outer"
).merge(cachaca_metrics_roberta_selected, on="split", how="outer")
```

```python
merged = merged_cleanconll.merge(merged_cachaca, on="split", how="outer")
```

```python
merged
```

```python
merged.columns
```

```python
split_order = ["std", "heur_len", "heur_rare", "advs", "loc", "semantic", "reverse"]

merged["split"] = merged["split"].astype(str).str.strip()

merged["split"] = pd.Categorical(merged["split"], categories=split_order, ordered=True)
merged = merged.sort_values("split", kind="stable").reset_index(
    drop=True
)
```

```python
accuracy_df = merged[["split"] + [c for c in merged.columns if "accuracy" in c]]
```
