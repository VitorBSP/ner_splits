```python
import numpy as np
import pandas as pd
```

```python
corpus = "ener"
ener_metrics_albert = pd.read_csv("ener/albert-base-v2_overall_metrics.csv")
name = "albert"
ener_metrics_albert_selected = ener_metrics_albert[
    ["split", "overall_accuracy", "f1_macro"]
].rename(
    columns={
        "overall_accuracy": f"overall_accuracy_{name}_{corpus}",
        "f1_macro": f"f1_macro_{name}_{corpus}",
    }
)

ener_metrics_bert = pd.read_csv(
    "ener/bert-base-cased_overall_metrics.csv"
)
name = "bert"
ener_metrics_bert_selected = ener_metrics_bert[
    ["split", "overall_accuracy", "f1_macro"]
].rename(
    columns={
        "overall_accuracy": f"overall_accuracy_{name}_{corpus}",
        "f1_macro": f"f1_macro_{name}_{corpus}",
    }
)

ener_metrics_roberta = pd.read_csv(
    "ener/xlm-roberta-base_overall_metrics.csv"
)
name = "roberta"
ener_metrics_roberta_selected = ener_metrics_roberta[
    ["split", "overall_accuracy", "f1_macro"]
].rename(
    columns={
        "overall_accuracy": f"overall_accuracy_{name}_{corpus}",
        "f1_macro": f"f1_macro_{name}_{corpus}",
    }
)

merged_ener = ener_metrics_bert_selected.merge(
    ener_metrics_albert_selected, on="split", how="outer"
).merge(ener_metrics_roberta_selected, on="split", how="outer")
```

```python
corpus = 'lener'
lener_metrics_albert = pd.read_csv("lener/albertina-base_overall_metrics.csv")
name = "albert"
lener_metrics_albert_selected = lener_metrics_albert[
    ["split", "overall_accuracy", "f1_macro"]
].rename(
    columns={
        "overall_accuracy": f"overall_accuracy_{name}_{corpus}",
        "f1_macro": f"f1_macro_{name}_{corpus}",
    }
)

lener_metrics_bert = pd.read_csv(
    "lener/bert-base-portuguese-cased_overall_metrics.csv"
)
name = "bert"
lener_metrics_bert_selected = lener_metrics_bert[
    ["split", "overall_accuracy", "f1_macro"]
].rename(
    columns={
        "overall_accuracy": f"overall_accuracy_{name}_{corpus}",
        "f1_macro": f"f1_macro_{name}_{corpus}",
    }
)

lener_metrics_roberta = pd.read_csv(
    "lener/xlm-roberta-base_overall_metrics.csv"
)
name = "roberta"
lener_metrics_roberta_selected = lener_metrics_roberta[
    ["split", "overall_accuracy", "f1_macro"]
].rename(
    columns={
        "overall_accuracy": f"overall_accuracy_{name}_{corpus}",
        "f1_macro": f"f1_macro_{name}_{corpus}",
    }
)
```

```python
merged_lener = lener_metrics_bert_selected.merge(
    lener_metrics_albert_selected, on="split", how="outer"
).merge(lener_metrics_roberta_selected, on="split", how="outer")
```

```python
merged = merged_lener.merge(merged_ener, on="split", how="outer")
```

```python
merged
```

```python
merged.columns
```

```python
split_order = ["std", "heur_len", "heur_rare", "advs", "loc", "semantic", "reverse"]

merged["split"] = merged["split"].astype(str).str.strip()

merged["split"] = pd.Categorical(merged["split"], categories=split_order, ordered=True)
merged = merged.sort_values("split", kind="stable").reset_index(
    drop=True
)
```

```python
accuracy_df = merged[["split"] + [c for c in merged.columns if "accuracy" in c]]
```
